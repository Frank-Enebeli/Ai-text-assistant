```python
from dotenv import load_dotenv
import os

load_dotenv()

api_key = os.getenv("OPENAI_API_KEY")

print("API key loaded:", api_key is not None)
```

```python
from openai import OpenAI

client = OpenAI(api_key=api_key)

print("OpenAI client created successfully.")
```

```python
response = client.responses.create(
    model="gpt-5-nano",
    input="Explain artificial intelligence in one simple sentence."
)

print(response.output_text)
```

```python
prompts = {
    "summarize": "Summarize the following text clearly and concisely:",
    
    "professional": "Rewrite the following text in a professional tone:",
    
    "key_points": "Extract the main points from the following text:",
    
    "quiz": "Create three questions and answers based on the following text:"
}

print(prompts)
```

creating a text field below to hold the text variable

```python
text = """
Artificial intelligence helps computers perform tasks that normally require
human intelligence. These tasks include understanding language, recognizing
images, making decisions, and solving problems.
"""
```

```python
task = "Quiz"

prompt = f"""
{prompts[task]}

{text}
"""

print(prompt)
```

```python

response = client.responses.create(
    model="gpt-5-nano",
    input=prompt
)

print(response.output_text)
```

```python
def transform_text(text, task):
    prompt = f"""
    {prompts[task]}

    {text}
    """

    response = client.responses.create(
        model="gpt-5-nano",
        input=prompt
    )

    return response.output_text
```

```python
result = result = transform_text(text, "key_points")

print(result)
```

```python
result = result = transform_text(text, "summarize")

print(result)
```
